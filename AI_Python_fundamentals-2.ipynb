-tuples are immutable

-defined by round brackets ()

-indexing of tuples works the same way as lists and strings

-tuples have an index and are hence ordered

-single element tuple is defined like this: (2, ) instead of just (2)

```python
my_tup = ("mango", 1, True, 3.12)
```

```python
my_tup[:2]
```

```python
my_tup[2] = False #immutable
```

```python
type([2])
```

```python
type((2))
```

```python
type((2,)) #single element tuple needs a comma at the end
```

```python
#to insert an element into a tuple, you have to make a whole new tuple and overwrite the previous one

my_tup = my_tup[:1] + (4.13,) + my_tup[1:]

my_tup
```

## **IF/ELIF/ELSE**

- you can have a standalone if statement, or an if combined with elif, or an if combined with else, or an if combined with both elif and else.

- if statement is the first check. If it evaluates to true, only the indented code under it will be executed and the remaining elif and else statements will be passed.

- in case the if statement evaluates to false, the indented code under it will be skipped and will jump to the next elif

- if that elif fails, then it skips the code under it and goes to next elif. and so on until else.

- At any point, if any elif evaluates to true, the next elifs and else will be skipped entirely

HENCE, ORDER OF CONDITIONS MATTERS!

```python
#program should check if num is even or odd

num = 11


if num % 2 == 0:

    print("num is even")
```

```python
num = 11


if num % 2 == 0:

    print("num is even")

else:

    print("num is not even")
```

#### **range() Function**

- range(start, stop, step) will generate integers starting from start, ending at stop (but not including stopping integer), and with appropriate steps, if step is also specified.

- If only one number is given to the range function, like range(5), it will generate numbers starting from zero up till (but not including) that number

```python
range(5)
```

```python
print(list(range(5)))
print(list(range(5, 10)))
print(list(range(2, 11, 2)))
```

```python
for i in [1,2,3,4,5]: #does not make sense to generate a list just to print python 5 times

    print("python")
```

```python
for i in range(5): #just use range when you want to run a loop a certain number of times

    print("python")
```

When running a loop on an iterable object, sometimes you need indices (for objects that have indices) and sometimes you don't

if we want to capitalize fruits in place, we need to iteratively repeat the following process.
So we need to generate indices in the loop

fruits[0] = fruits[0].capitalize() <br>
fruits[1] = fruits[1].capitalize() <br>
fruits[2] = fruits[2].capitalize() <br>


if we want to capitalize each fruit and push it into a separate list, we don't need indices. we can just grab whole fruits.

```python
#edit list in place. we need indices

fruits = ["apple", "orange", "banana", "melon", "kiwi"]


for i in range(len(fruits)):

    fruits[i] = fruits[i].capitalize()


fruits
```

```python
#push into a separate list. we don't need indices.

fruits = ["apple", "orange", "banana", "melon", "kiwi"]
capital_fruits = []


for item in fruits:

    capital_fruits.append(item.capitalize())


capital_fruits
```

##### **nested loops**

```python
fruits = ["apple", "orange", "banana", "melon", "kiwi"]


for item in fruits:

    for letter in item:

        print(letter)

    print("\n")
```

##### **break, continue and pass statements in for loops**

- A break statement terminates the for loop. No code inside the loop is executed after break
- A continue statement skips to the next iteration without executing the remaining code in that iteration
- A pass statement does nothing. It can act as a placeholder for code.

```python
fruits = ["apple", "orange", "banana", "melon", "kiwi"]


for item in fruits:

    if item == "melon":
        break

    print(item)

```

```python
fruits = ["apple", "orange", "banana", "melon", "kiwi"]


for item in fruits:


    if item == "melon":
        continue

    print(item)
```

```python
fruits = ["apple", "orange", "banana", "melon", "kiwi"]


for item in fruits:

    pass
```

### **WHILE LOOP**

- A while loop will keep running as long as a condition is true
- It will terminate if the condition is false or if it reaches a break statement inside it
- You can use break and continue inside while loops the same way you use them inside for loops

```python
i = 6


while i > 0:

    print(i)
```

```python
i = 6


while i > 0:

    print(i)
    i -= 1
```

#### **Importance of the .copy() method**

with mutable structures like lists, dictionaries and dataframes (in the pandas library), if you set b = a, any changes to a will also propagate to b. If you don't want this to happen, set b = a.copy()

```python
#without .copy()

a = [1,2,3,4]
b = a

print("before")
print("a:", a)
print("b:", b)


print("\n")


a.append(5) #just making a change to a

print("after")
print("a:", a)
print("b:", b)
```

```python
#with .copy()

a = [1,2,3,4]
b = a.copy()

print("before")
print("a:", a)
print("b:", b)


print("\n")


a.append(5) #just making a change to a

print("after")
print("a:", a)
print("b:", b)
```
