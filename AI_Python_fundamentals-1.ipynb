### **OPERATORS (MATHEMATICAL, COMPARISON, LOGICAL AND ASSIGNMENT)**

MATHEMATICAL

```python
x = 10
y = 20


print(x + y) #add
print(x - y) #subtract
print(x * y) #multiply
print(x / y) #divide
print(x ** 4) #exponentiation
print(x % 4) #modulo
print(x // 4) #floor division
```

```python
9999//100
```

COMPARISON

```python
x = 10
y = 20


print(x > y) #greater than
print(x < y) #less than
print(x <= y) #less than or equal to
print(x >= y) #greater than or equal to
print(x == 10) #equality
print(x != 10) #inequality
```

LOGICAL

```python
x = 10
y = 20

print(x < 12 and y == 20 and x != y) #and: all conditions should be true to get true overall
print(x > 12 or y == 20 or x == y) #or: one condition should be true to get true overall
print(not x == 10) #not: makes True into False and vice versa
```

### **STRINGS**

-Strings are immutable (unchangeable)

-Defined by quotes (single or double)

-Strings have an index and are hence ordered

```python
my_str = "abcdefghijklmnopqrstuvwxyz"
```

```python
print(type(my_str))
print(len(my_str))
```

#### **String Indexing/Slicing**

-indexes start from 0

-string[start:stop:step]

-starting index is included in the slice but stopping index is not

-starting index needs to be bigger than stopping index and step needs to be negative if you are slicing in reverse

-a step of 2 means skip every other element in the slice. a step of 3 means skip two elements

```python
my_str[12]
```

```python
my_str[3]
```

```python
my_str = "abcdefghijklmnopqrstuvwxyz"

print(my_str[3])
print(my_str[2:11])
print(my_str[:4])
print(my_str[23:])
print(my_str[1:11])
print(my_str[1:11:2])
print(my_str[1:11:3])
print(my_str[-3])
print(my_str[23:10:-1])
print(my_str[::-1])
```

### **LISTS**

-Lists are mutable (changeable)

-Defined by square brackets [ ]

-Lists have an index and are hence ordered

```python
my_list = ["apple", "orange", "banana", "kiwi", "watermelon"]
```

#### **List Indexing/Slicing**

```python
print(my_list[2])
print(my_list[1:4])
print(my_list[1:4:2])
print(my_list[-1])
print(my_list[-1][3])
```

#### **List Item Assignment**


```python
my_list[2] = "grape"

my_list
```

#### **List methods**

-List methods can make direct changes to the original list. So they will DIRECTLY alter your list without having to reassign the result back to the list

https://www.w3schools.com/python/python_ref_list.asp

```python
my_list.insert(2, "melon")

my_list
```

```python
my_list.count("orange")
```

```python
my_list.append("lemon")

my_list
```
